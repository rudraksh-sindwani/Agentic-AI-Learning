# Authorized Imports in AI Agents

In this notebook, I explore how a `CodeAgent` can be given permission to import additional Python modules.

I use the `datetime` module to allow the agent to work with the current time and calculate when a series of tasks will be completed.

In this notebook, I will learn how to:

- Create a `CodeAgent` without external tools
- Use `additional_authorized_imports`
- Allow an agent to use the `datetime` module
- Let the agent perform time-based calculations using Python

## 1. Creating the Agent

In this example, the agent does not need any external tools. Instead, I allow it to import Python's `datetime` module using `additional_authorized_imports`.

This allows the agent to access the current time and perform time-based calculations.

```python
from smolagents import CodeAgent, InferenceClientModel

model = InferenceClientModel(
    model_id="Qwen/Qwen2.5-Coder-32B-Instruct"
)

agent = CodeAgent(
    tools=[],
    model=model,
    additional_authorized_imports=["datetime"]
)
```

## 2. Running the Agent

Now I give the agent a list of preparation tasks with their durations.

The agent can use Python code and the authorized `datetime` module to calculate the total preparation time and determine when the party will be ready.

```python
agent.run(
    """
    Alfred needs to prepare for the party. Here are the tasks:
    
    1. Prepare the drinks - 30 minutes
    2. Decorate the mansion - 60 minutes
    3. Set up the menu - 45 minutes
    4. Prepare the music and playlist - 45 minutes

    If we start right now, at what time will the party be ready?
    """
)
```
