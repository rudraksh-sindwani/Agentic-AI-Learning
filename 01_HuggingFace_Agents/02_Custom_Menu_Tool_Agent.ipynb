# Custom Menu Tool Agent

In this notebook, I build an AI agent with a custom tool using the Hugging Face `smolagents` library.

The agent can use a custom Python function to suggest a menu based on the type of occasion.

In this notebook, I will learn how to:

- Create a custom tool using the `@tool` decorator
- Define the tool's inputs and purpose
- Give a custom tool to a `CodeAgent`
- Use an LLM as the agent's reasoning engine
- Understand how an agent decides when to use a tool

## 1. Importing Libraries and Creating the Model

First, I import the required components from `smolagents`. I use `InferenceClientModel` to connect the agent to a language model, which acts as the agent's reasoning engine.

```python
from smolagents import CodeAgent, tool, InferenceClientModel

model = InferenceClientModel(
    model_id="Qwen/Qwen2.5-Coder-32B-Instruct"
)
```

## 2. Creating a Custom Tool

Here, I create a custom tool called `suggest_menu` using the `@tool` decorator.

The tool accepts an occasion as input and returns a suitable menu. The docstring provides the agent with information about the tool's purpose and expected input.

```python
@tool
def suggest_menu(occasion: str) -> str:
    """
    Suggests a menu based on the occasion.

    Args:
        occasion (str): The type of occasion for the party. Allowed values are:
                        - "casual": Menu for a casual party.
                        - "formal": Menu for a formal party.
                        - "superhero": Menu for a superhero-themed party.
                        - "custom": A custom menu.
    """
    if occasion == "casual":
        return "Pizza, snacks, and drinks."
    elif occasion == "formal":
        return "3-course dinner with wine and dessert."
    elif occasion == "superhero":
        return "Buffet with high-energy and healthy food."
    else:
        return "Custom menu for the butler."
```

## 3. Creating the Agent

Now I create a `CodeAgent` and provide it with the `suggest_menu` tool.

The agent uses the language model as its reasoning engine and can decide when to call the custom tool.

```python
agent = CodeAgent(
    tools=[suggest_menu],
    model=model
)
```

## 4. Running the Agent

Now I give the agent a task to prepare a formal menu. The agent can understand the request and decide to use the `suggest_menu` tool with `"formal"` as the input.

agent.run("Prepare a formal menu for the party.")

## 5. What I Learned

In this notebook, I learned how to create and use a custom tool with an AI agent.

- The `@tool` decorator converts a Python function into a tool that can be provided to an agent.
- The function defines the actual logic of the tool.
- The type hints describe the expected input and output.
- The docstring explains the tool's purpose and expected input to the agent.
- `CodeAgent` uses the language model as its reasoning engine.
- The agent can decide when it is appropriate to call an available tool.
- `agent.run()` gives the agent a task to complete.

I also observed that an agent may answer using its language model instead of calling a tool if the available tool is not suitable for the requested task.
